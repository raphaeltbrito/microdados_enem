```python
import pandas as pd
```

```python
ANO = 2009
arquivo_microdados = f"MICRODADOS_ENEM_{ANO}.csv"
```

```python
colunas_microdados = ["TP_SEXO", "Q3", "TP_FAIXA_ETARIA", "SG_UF_PROVA",
            "TP_ST_CONCLUSAO", "NU_ANO", "Q17", "Q18", "Q21", "NU_NOTA_CN", "NU_NOTA_CH", "NU_NOTA_LC", "NU_NOTA_MT", "NU_NOTA_REDACAO"]
```

```python
microdados = pd.read_csv(arquivo_microdados, sep=";", encoding="latin-1", usecols=colunas_microdados)
```

```python
microdados.head()
```

```python
microdados_limpos = microdados.dropna()
```

```python
microdados_limpos.head()
```

```python
microdados_limpos = microdados_limpos.rename(columns={
    "Q3" : "TP_COR_RACA",
    "Q17" : "Q001",
    "Q18" : "Q002",
    "Q21" : "Q006"
    
})
```

```python
microdados_limpos.head()
```

```python
microdados_limpos.to_csv(f"microdados_{ANO}_limpo.csv", sep=";", encoding="latin-1", index=False)
```
