```python
import pandas as pd
```

```python
ANO = 2022
arquivo_microdados = f"microdados_enem_{ANO}.csv"
```

```python
colunas_microdados = ["TP_SEXO","TP_COR_RACA","TP_FAIXA_ETARIA","SG_UF_PROVA","Q006","TP_ESCOLA",
            "TP_ST_CONCLUSAO","NU_ANO","Q001","Q002","Q003","Q004", "NU_NOTA_CN","NU_NOTA_CH","NU_NOTA_LC","NU_NOTA_MT","NU_NOTA_REDACAO"] 
```

```python
microdados = pd.read_csv(arquivo_microdados, sep=";", encoding="latin-1", usecols=colunas_microdados)
```

```python
microdados.head()
```

```python
print(microdados.isna().sum())
```

```python
microdados_limpos = microdados.dropna()
```

```python
print(len(microdados), "->", len(microdados_limpos))
```

```python
microdados_limpos.to_csv(f"microdados_{ANO}_limpo.csv", index = False)
```
