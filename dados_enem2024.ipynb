```python
import pandas as pd
```

```python
ANO = 2024
arquivo_participantes = f"PARTICIPANTES_{ANO}.csv"
arquivo_resultados = f"RESULTADOS_{ANO}.csv"
```

```python
colunas_participantes = ["TP_SEXO","TP_COR_RACA","TP_FAIXA_ETARIA","SG_UF_PROVA","Q007","Q023",
            "TP_ST_CONCLUSAO","NU_ANO","Q001","Q002","Q003","Q004"]   

colunas_resultados = ["NU_NOTA_CN","NU_NOTA_CH","NU_NOTA_LC","NU_NOTA_MT","NU_NOTA_REDACAO"] 
```

```python
participantes = pd.read_csv(arquivo_participantes, sep=";", encoding="latin-1", usecols=colunas_participantes)
resultados = pd.read_csv(arquivo_resultados, sep=";", encoding="latin-1", usecols=colunas_resultados)
```

```python
participantes.head()
```

```python
resultados.head()
```

```python
print(participantes.isna().sum())
print(resultados.isna().sum())
```

```python
participantes_limpo = participantes.dropna()
resultados_limpo = resultados.dropna()
```

```python
print(len(participantes), "->", len(participantes_limpo))
print(len(resultados), "->", len(resultados_limpo))
```

```python
participantes_limpo.to_csv(f"participantes_{ANO}_limpo.csv",index=False)
resultados_limpo.to_csv(f"resultados_{ANO}_limpo.csv",index=False)
```
