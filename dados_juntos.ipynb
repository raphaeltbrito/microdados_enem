```python
import pandas as pd
```

```python
df_2009 = pd.read_csv("microdados_2009_limpo.csv", sep=";", encoding="latin-1")
df_2013 = pd.read_csv("microdados_2013_limpo.csv", sep=";", encoding="latin-1")
df_2016 = pd.read_csv("microdados_2016_limpo.csv", sep=";", encoding="latin-1")
df_2017 = pd.read_csv("microdados_2017_limpo (1).csv", sep=",", encoding="latin-1")
df_2018 = pd.read_csv("microdados_2018_limpo (1).csv", sep=",", encoding="latin-1")
df_2019 = pd.read_csv("microdados_2019_limpo.csv", sep=",", encoding="latin-1")
df_2020 = pd.read_csv("microdados_2020_limpo.csv", sep=",", encoding="latin-1")
df_2021 = pd.read_csv("microdados_2021_limpo.csv", sep=",", encoding="latin-1")
df_2022 = pd.read_csv("microdados_2022_limpo.csv", sep=",", encoding="latin-1")
df_2023 = pd.read_csv("microdados_2023_limpo.csv", sep=",", encoding="latin-1")

dfs = [df_2009, df_2013, df_2016, df_2017, df_2018, df_2019, df_2020, df_2021, df_2022, df_2023]
```

```python
colunas_comuns = [col for col in df_2009.columns if all(col in df.columns for df in dfs)]
```

```python
colunas_exclusivas = []
for df in dfs:
    for col in df.columns:
        if col not in colunas_comuns and col not in colunas_exclusivas:
            colunas_exclusivas.append(col)
```

```python
ordem_final = colunas_comuns + colunas_exclusivas
```

```python
df_unificado = pd.concat(dfs, ignore_index=True)

ordem_colunas = [
    'NU_ANO', 'SG_UF_PROVA', 'TP_SEXO', 'TP_FAIXA_ETARIA', 'TP_COR_RACA', 'TP_ESCOLA', 'TP_ST_CONCLUSAO',
    'NU_NOTA_LC', 'NU_NOTA_REDACAO', 'NU_NOTA_CH', 'NU_NOTA_CN', 'NU_NOTA_MT',
    'Q001', 'Q002', 'Q003', 'Q004', 'Q006'
]

df_unificado = df_unificado[ordem_colunas]
```
