```python
import pandas as pd
```

```python
df_2009 = pd.read_csv("microdados_2009_limpo.csv", sep=";", encoding="latin-1")
df_2013 = pd.read_csv("microdados_2013_limpo.csv", sep=";", encoding="latin-1")
df_2016 = pd.read_csv("microdados_2016_limpo.csv", sep=";", encoding="latin-1")
df_2020 = pd.read_csv("microdados_2020_limpo.csv", sep=",", encoding="latin-1")
df_2021 = pd.read_csv("microdados_2021_limpo.csv", sep=",", encoding="latin-1")
df_2022 = pd.read_csv("microdados_2022_limpo.csv", sep=",", encoding="latin-1")

dfs = [df_2009, df_2013, df_2016, df_2020, df_2021, df_2022]
```

```python
colunas_comuns = [col for col in df_2009.columns if all(col in df.columns for df in dfs)]
```

```python
colunas_exclusivas = []
for df in dfs:
    for col in df.columns:
        if col not in colunas_comuns and col not in colunas_exclusivas:
            colunas_exclusivas.append(col)
```

```python
ordem_final = colunas_comuns + colunas_exclusivas
```

```python
df_unificado = pd.concat(dfs, ignore_index=True)
df_unificado = df_unificado[ordem_final]
```
