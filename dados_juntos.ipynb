```python
import pandas as pd
```

```python
df_2009 = pd.read_csv("microdados_2009_limpo.csv", sep=";", encoding="latin-1")
df_2013 = pd.read_csv("microdados_2013_limpo.csv", sep=";", encoding="latin-1")
df_2016 = pd.read_csv("microdados_2016_limpo.csv", sep=";", encoding="latin-1")

dfs = [df_2009, df_2013, df_2016]
```

```python
colunas_comuns = [col for col in df_2009.columns if all(col in df.columns for df in dfs)]
```

```python
colunas_2009_exc = [col for col in df_2009.columns if col not in colunas_comuns]
colunas_2013_exc = [col for col in df_2013.columns if col not in colunas_comuns and col not in colunas_2009_exc]
colunas_2016_exc = [col for col in df_2016.columns if col not in colunas_comuns and col not in colunas_2009_exc and col not in colunas_2013_exc]
```

```python
ordem_final = colunas_comuns + colunas_2009_exc + colunas_2013_exc + colunas_2016_exc
```

```python
df_unificado = pd.concat(dfs, ignore_index=True)
df_unificado = df_unificado[ordem_final]
```
